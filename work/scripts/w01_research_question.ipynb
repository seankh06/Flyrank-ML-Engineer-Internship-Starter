## 1. My lane (or freestyle) and why

**Ranking Signal Analysis**

I'm choosing the Ranking Signal Analysis lane. From the previous discoveries, I found that search volume doesn't really affect impressions, and that position has a huge effect on CTR. This was really interesting and made me wonder about the correlation and relationship between
word count of an article and its CTR and sessions over the last 90 days. Understanding this could give insight into whether writing longer content is actually worth the effort, or whether it makes little difference to how much a page actually gets clicked and visited.

## 2. The question: decision, action, cost of a wrong call

This analysis could help editors and content strategists decide the ideal word count range for future articles, based on what actually correlates with better CTR and sessions, instead of guessing.

If the recommendation is wrong in one direction, for example suggesting writers spend extra time on longer articles when length doesn't actually matter, it costs time, effort, and money that could have gone elsewhere.

If it's wrong in the other direction, recommending shorter articles when length actually does matter, the page ends up under-optimized, and the business loses out on potential traffic, engagement, or revenue it could have captured.

## 3. Quick look at the data (2-3 real numbers)


The correlation between word_count and ctr is -0.119, a very weak negative relationship, basically close to no relationship at all. Longer articles don't seem to meaningfully affect whether someone clicks from the search results. However, word_count and sessions_90d show a moderate positive correlation of 0.332, meaning longer articles tend to get noticeably more
sessions. This is an interesting contrast: article length barely affects the initial click decision, but it does seem to relate to how much a page gets visited afterward. This gap between the two metrics is worth digging into further over the next 7 weeks, since it suggests word count might influence engagement or return visits more than it influences the first click.

```python
import pandas as pd
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

correlation_1 = df['word_count'].corr(df['ctr'])
correlation_2 = df['word_count'].corr(df['sessions_90d'])

print(f"Correlation (word_count vs ctr): {correlation_1:.3f}")
print(f"Correlation (word_count vs sessions_90d): {correlation_2:.3f}")
```

## 4. Careful words: what I can and can't claim

This analysis is observational and decision-support only. I can say that word_count shows a weak correlation with CTR and a moderate correlation with sessions_90d within this specific dataset, and use that pattern to suggest a general word count range for editors to consider. This is directional evidence, not a guarantee, since correlation only shows that two things move together, not that one causes the other.

I cannot claim that writing longer articles causes more sessions, since other factors (like content quality, topic, or search intent) could be driving both word count and sessions at the same time. I also cannot claim this holds true for every type of content or every client, since this is based on one anonymized dataset at one point in time. Finally, this analysis has nothing to do with how Google's search algorithm actually works. It only describes patterns observed in this dataset, not the underlying mechanics of search ranking.
